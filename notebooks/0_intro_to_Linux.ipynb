Search for some online Linux tutorials or command cheatsheets, for example

__Navigation__
```bash
pwd                 # Show current directory
ls                  # List files
ls -la              # List all files with details
cd /path/to/dir     # Change directory
cd ..               # Go up one directory
cd ~                # Go to home directory
```


__Files and Directories__
```bash
touch file.txt              # Create an empty file
mkdir mydir                 # Create a directory
cp file.txt backup.txt      # Copy a file
cp -r mydir backup          # Copy a directory
mv file.txt newname.txt     # Move or rename
rm file.txt                 # Delete a file
rm -r mydir                 # Delete a directory
```

__Viewing Files__
```bash
cat file.txt                # Display file contents
less file.txt               # View file page by page
head file.txt               # Show first 10 lines
tail file.txt               # Show last 10 lines
```

__Searching__
```bash
find . -name "*.txt"        # Find .txt files
grep "error" file.txt       # Search for text
grep -r "error" .           # Search recursively
which python                # Find location of a command
```


__Permissions__
```bash
ls -l                       # View permissions
chmod +x script.sh          # Make a script executable
chmod 644 file.txt          # Set file permissions
chown user file.txt         # Change file owner
```

__Processes__
```bash
ps aux                      # List running processes
top                         # Interactive process viewer
kill 1234                   # Stop process by PID
kill -9 1234                # Force-kill process
```


__Disk and Memory__
```bash
df -h                       # Show disk space
du -sh .                    # Show directory size
free -h                     # Show memory usage
```

__Archives__
```bash
tar -czf backup.tar.gz mydir/       # Create compressed archive
tar -xzf backup.tar.gz              # Extract archive
zip -r backup.zip mydir/            # Create ZIP
unzip backup.zip                    # Extract ZIP
```

__Package Management__

```bash
sudo apt update                     # Update package lists
sudo apt upgrade                    # Upgrade packages
sudo apt install package_name       # Install a package
sudo apt remove package_name        # Remove a package
```

## 3. Text editor

Learn to use at least one type of text editor, e.g. `vi`, `vim`, `emacs`, or `nano`, with which you can edit plain text files within a Linux terminal (You can always use other graphic-interface text editors such as Jupyter to view or modify text files, but it won't be very efficient, especially for large files and batch tasks). 


### Basic `vi` commands

__Mode__

| Mode             | How to enter  | Purpose                      |
| ---------------- | ------------- | ---------------------------- |
| **Normal**       | `Esc`         | Navigation and commands      |
| **Insert**       | `i`, `a`, `o` | Type/edit text               |
| **Command-line** | `:`           | Save, quit, search, settings |


__Navigation__
| Command   | Description       |
| --------- | ----------------- |
| `h` / `←` | Move left         |
| `j` / `↓` | Move down         |
| `k` / `↑` | Move up           |
| `l` / `→` | Move right        |
| `0`       | Beginning of line |
| `$`       | End of line       |
| `gg`      | Beginning of file |
| `G`       | End of file       |
| `w`       | Next word         |
| `b`       | Previous word     |
| `Ctrl-f`  | Page down         |
| `Ctrl-b`  | Page up           |

__Insert and Edit__
| Command  | Description           |
| -------- | --------------------- |
| `i`      | Insert before cursor  |
| `a`      | Insert after cursor   |
| `o`      | Open new line below   |
| `O`      | Open new line above   |
| `x`      | Delete character      |
| `dd`     | Delete current line   |
| `dw`     | Delete word           |
| `D`      | Delete to end of line |
| `u`      | Undo                  |
| `Ctrl-r` | Redo                  |

__Copy and Paste__
| Command | Description              |
| ------- | ------------------------ |
| `yy`    | Copy (yank) current line |
| `yw`    | Copy word                |
| `p`     | Paste after cursor       |
| `P`     | Paste before cursor      |


__Search__
| Command    | Description     |
| ---------- | --------------- |
| `/pattern` | Search forward  |
| `?pattern` | Search backward |
| `n`        | Next match      |
| `N`        | Previous match  |

__Save and Quit__
| Command | Description         |
| ------- | ------------------- |
| `:w`    | Save                |
| `:q`    | Quit                |
| `:wq`   | Save and quit       |
| `:q!`   | Quit without saving |
| `:x`    | Save and quit       |


__Search and Replace__
| Command | Description         |
| ------- | ------------------- |
| `:%s/old/new/g`    | Replace all occurrences of old with new throughout the file.                |    

__Other settings__
| Command       | Description          |
| ------------- | -------------------- |
| `:e filename` | Open a file          |
| `:set nu`     | Show line numbers    |
| `:set nonu`   | Hide line numbers    |
| `:help`       | Open help            |
| `:help w`     | Help for command `w` |



__Command Workflow__

```
vi filename
      ↓
    Press i
      ↓
   Edit text
      ↓
    Press Esc
      ↓
     :wq
      ↓
      Enter
```
