## 1.  Installation and virtual environment setup

Once Python 3 is properly installed, it is recommended to create a dedicated virtual environment for Jupyter Notebook and future Python package installations. This helps keep project dependencies isolated and avoids conflicts between packages.

__References:__


https://packaging.python.org/en/latest/guides/installing-using-pip-and-virtual-environments/#create-and-use-virtual-environ

Summary of Command-Line Steps  (e.g. in Linux Terminal) for Setting Up a Python Virtual Environment and Installing Jupyter and Python Packages


```bash
# create a new virtual environment in a local folder named .venv  
python3 -m venv .venv 

# activate virtual environment 
source .venv/bin/activate 

# check if python command is linked to .venv/bin/python 
which python 

# (optional) upgrade pip
python -m pip install --upgrade pip

# (optional) verify environment
python --version
pip --version

# (optional) check current packages 
python -m pip list 

# install jupyter 
pip install jupyter 

# install Python packages, e.g.
pip install numpy pandas matplotlib scikit-learn

# (optional) create a new ipython kernel for jupyter named .venv 
python -m ipykernel install --user --name=.venv 

# exit virtual environment 
deactivate 
```

There two types of cells in Jupyter notebook: __Code__ cell or __Markdown__ cell. Both of them need to be run to see the effect (output).

| Shortcut | Action |
|---|---|
| `Esc → M` | Change current cell into a Markdown cell |
| `Esc → Y` | Change current cell into a Code cell |
| `Esc → A` | Insert cell above |
| `Esc → B` | Insert cell below |
| `Esc → D, D` | Delete cell |
| `Shift + Enter` | Run cell and go to next cell|
| `Ctrl + Enter` | Run cell and stay in current cell|

### 3.1 File management

How to load and save files within Colab on Google Drive? (You will need to do it one day.)

By default, your Colab *.ipynb file should be created and saved on your Google Drive under the folder "My Drive > Colab Notebooks"

You should first mount the Google Drive using the following command. You will be asked to choose your Google account, sign in and click continue.

(The following sample Python code should be run on Google Colab)

```python
# Sign in and authorization of Google account here
from google.colab import drive
drive.mount('/content/drive')
```

You can use the following command to list files and folders under your current Colab Notebooks folder (This step is not necessary, but it is always good to check the file you need is there)


```python
# show contents under the Colab Notebooks folder
!ls "/content/drive/My Drive/Colab Notebooks"
```

When you want to load or read a data file on Colab, you should first upload that file onto your Google Drive, also under the same folder My Drive > Colab Notebooks. For example, if the file is ```mydatafile.txt``` and you want to load it into a numpy array, the command is like

```python
my_array = np.loadtxt("/content/drive/My Drive/Colab Notebooks/mydatafile.txt") # load a txt file if it exits under this path

```

When your Python codes generate and save a file, it should be on your Google Drive under the same folder My Drive > Colab Notebooks. You could choose to manually download it for futher usage. You could also use the following command to automatically download it (say an image file ```my_image.png```)

```python
from google.colab import files
files.download("my_image.png") # download my_image.png (if it exists)
```
