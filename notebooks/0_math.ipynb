# Mathematical Preparation: Linear Algebra, Calculus, Probability etc.


Kai Zhang, The University of Texas at Tyler

The mathematical minimum to start with basic machine learning: 

- linear algebra: vectors, matrices, and tensors -- how to add and multiply them

- calculus: derivatives, chain rule, gradients -- to minimize loss function

- probability and statistics: probability distribution, conditional probability, Bayes' rule, maximum likelihood estimation, sampling -- to do generative modelling
