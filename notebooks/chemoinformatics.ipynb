## SMILES code

In a Simplified Molecular Input Line Entry System (SMILES) string, e.g. `CC(=O)Nc1ccc(O)cc1`

* atom represented by element symbol letter, e.g. C, N, O, etc.
* hydrogen usually ignored
* single bond not displayed
* double, triple, quadruple bond uses =, #, $
* bonded atoms stand nearby
* rings are broken at an arbitrary point; numerical labels are added at non-adjacent atoms where rings are broken
* aromatic atoms are in lower case, e.g. c, n, b, etc.
* side chains are enclosed in ()

same molecule may be written in different SMILES string
